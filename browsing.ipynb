# Overview
The asset system of Anno is very nested: Every asset has a template. The template defines the properties and some default values. Properties are like Python dictinaries with names as keys and values (which can be other properties or attributes like strings, ints, references). Properties can be reused in multiple templates and are thus stored in a seperate `properties.xml`. The asset extractor puts together all this information so that you can see all attributes an asset has, no matter over how many steps these are inherited.
Print all templates. Click on `scrollable element` after executing the cell to see the full text.

```python
templates.print_tree()
```

Get a specific template and show its structure - this shows all inherited and default properties (more than one directly sees in the `templates.xml`):

```python
template_farm = templates["Production Field"]

# uncomment the following if you want to explore the assets for Anno 1800
# template_farm = templates["FarmBuilding"].print_tree()

template_farm.print_tree()
```

The `properties.xml` contains additional description texts for most attributes. To show them call the `print_meta_tree` function on any asset, template or property:

```python
template_farm.print_meta_tree()
```

Show the GUIDs of all assets of template FarmBuilding

```python
template_farm.assets
```

You can now get the asset by calling

```python
farm = template_farm.assets[0]
```

All assets and properties can be treated like dictionaries. So you can get, e.g., the `Cost` property of the asset by calling `assets[1010262]["Cost"]` or `assets[1010262].Cost`. And you can chain it:

```python
farm.Cost.Costs.print_tree()
```

Since we now have a list here, we need the index access to go further:

```python
farm.Cost.Costs[0].print_tree()
```

Coins is a reference to another asset. We can inspect it by going the full path and doing '()' at the attribute to get its value:

```python
farm.Cost.Costs[0].Ingredient()
```

```python
farm.Cost.Costs[0].Ingredient().print_tree()
```

Like dicts, you can iterate assets (or properties) or check whether they contain a certain key:

```python
for property in farm:
    print(property.name)
```

Other object attributes methods available on templates, assets, properties, and attributes:
* `print_tree()` The printed tree is not identical to the xml as some nodes are skipped (and their children embedded into the parent; see property_path) and inherited values added.
* `print_meta_tree()`
* `name`
* `property_path` returns a string of all parent elements (including an index if its a list element). The path can be input in another asset to get its corresponding attribute. Note that the path is not identical to the sequence of xml nodes in the original xml file. For technical reasons, the `Values` node in templates and `Item` node of lists are skipped.
* `is_compound` if true, you can iterate the class to get its children, if false the class at hand is an `Attribute`
* `__class__.name`bultin method to get the class name of an object
* `__dict__` see all attributes defined on an object

Special attributes:
* `template`: Gives the template on `Asset` and `TemplateAttribute`.
* `text`: Gives a `Text` if there is an UI text available for an asset or property.
* `instances`: Dictionary of assets that inherit from the template/asset
* `referenced_by` Dictinary of other assets referencing this one (keys are GUIDs)
* `named_reference_collections`: A dictionary. Keys are strings, values are, e.g. `instances` or `referenced_by`.
