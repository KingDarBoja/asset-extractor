# Render single asset - for interactively modifying and updating template

```python
asset = list(templates["PopulationLevel"].instances.values())[0].source
```

```python
asset = assets[3185]
```

```python

# Set up Jinja2 environment
template_dir = Path("assetextractor/conversion/assetbrowser/templates")
env = Environment(loader=FileSystemLoader(str(template_dir)))
template = env.get_template("asset.html")

output_dir = Path("results/assetbrowser/assets").resolve()
output_dir.mkdir(parents=True, exist_ok=True)
```

```python
from html import escape

html = template.render(
    asset=asset,
    xml=escape(tostring(asset.node, pretty_print=True, encoding="unicode"))
        .replace('\n', '<br>').replace('  ', '&nbsp;&nbsp;')
)
asset_id = asset.safe_identifier
filename = f"{asset_id}.html"
filepath = output_dir / filename
with filepath.open("w", encoding="utf-8") as f:
    f.write(html)
```

```python
# Set up Jinja2 environment
template_dir = Path("assetextractor/conversion/assetbrowser/templates")
env = Environment(loader=FileSystemLoader(str(template_dir)))
template = env.get_template("template.html")

output_dir = Path("results/assetbrowser/templates").resolve()
output_dir.mkdir(parents=True, exist_ok=True)
```

```python
html = template.render(template=asset.template)
template_id = asset.template.name
filename = f"{template_id}.html"
filepath = output_dir / filename
with filepath.open("w", encoding="utf-8") as f:
    f.write(html)
```

```python
# Set up Jinja2 environment
template_dir = Path("assetextractor/conversion/assetbrowser/templates")
env = Environment(loader=FileSystemLoader(str(template_dir)))
template = env.get_template("template_cache.html")

output_dir = Path("results/assetbrowser").resolve()
output_dir.mkdir(parents=True, exist_ok=True)

html = template.render(cache=assets.templates)

filename = f"index.html"
filepath = output_dir / filename
with filepath.open("w", encoding="utf-8") as f:
    f.write(html)
```
