# Basic Setup - Loading Assets and Navigation

This notebook shows the basics of loading the Anno 117 asset data and navigating templates.

## Step 1: Load the Asset Cache

The AssetCache contains all game data: buildings, items, buffs, etc.

```python
from assetextractor.extraction.utils import Config
from assetextractor.parsing.core.assets import AssetCache

# Load the asset cache from config.json
config = Config.from_json("config.json")
assets = AssetCache.load(config)

print("Assets loaded successfully!")
print(f"Total assets: {len(assets.elements)}")
```

## Step 2: Explore Templates

Templates are categories of assets (like "Item", "BuildingBuff", "ProductionBuilding").

```python
# Show all available templates
print("Available templates:")
for template_name in sorted(assets.templates.elements.keys())[:20]:  # Show first 20
    template = assets.templates[template_name]
    print(f"  {template_name}: {len(template.assets)} assets")
```

## Step 3: Access Specific Assets

You can access assets by GUID or iterate through a template.

```python
# Access by GUID
item = assets.get(80510)  # Zorsines, Sarmatian Swordshaper
print(f"Asset by GUID: {item.guid}")
print(f"  Template: {item.template.name}")
print(f"  Name: {item.text.values.get('english', 'N/A') if item.text else 'N/A'}")
```

```python
# Iterate through items template (show first 5)
items_template = assets.templates["Item"]
print(f"\nFirst 5 items from {len(items_template.assets)} total:")

for i, item in enumerate(list(items_template.assets)[:5]):
    name = item.text.values.get('english', 'N/A') if item.text else 'N/A'
    print(f"  {item.guid}: {name}")
```

## Step 4: Navigate Asset Properties

Assets have properties organized hierarchically. Use `.find()` to navigate paths.

```python
# Get item properties
item = assets.get(80510)

# Access properties using dot notation
rarity = item.Item.Rarity() if hasattr(item, 'Item') else None 
# or
rarity = item.find("Item.Rarity")()
print(f"Rarity (literal): {rarity}")

# Or use find() for dynamic paths
trade_price = item.find("Item.TradePrice")
print(f"Trade Price: {trade_price() if trade_price else 'N/A'}")
```

## Step 5: Print Asset Tree

Use `print_tree()` to see all properties and values of an asset.

```python
# Print complete asset structure (this will be long!)
# Uncomment to see the full tree:
# item.print_tree()

# Instead, let's print just the Item property
if hasattr(item, 'Item'):
    item.Item.print_tree()
```
